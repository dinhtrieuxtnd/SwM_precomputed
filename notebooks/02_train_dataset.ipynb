```python
from src.utils.io import load_pickle
```

```python
from src.data.preprocessing import build_training_pair, build_negative_sequence
```

```python
sample = train_samples[6]
history = sample["history"]
target = sample["target"]
```

```python
input_ids, positive_ids = build_training_pair(history, target, 50, 0)
```

```python
user_items = set(history + [target])
mapping_keys = list(news_vector_mapping.keys())
negative_ids = build_negative_sequence(positive_ids, user_items, mapping_keys, 0)
```

```python
from src.data.mapping import map_id_to_vector
```
